```python
import pandas as pd
```

```python
df = pd.read_csv('data.csv')
df.drop(columns=['vessel_name', 'imo', 'call_sign', 'cargo', 'transceiver'], inplace=True)
df.head()
```

```python
# houston/galveston port
latitude_bounds = [28.5, 30.5]
longitude_bounds = [-96.5, -94.0]

df_houston = df[
    df['latitude'].between(*latitude_bounds) &
    df['longitude'].between(*longitude_bounds)
]

df_houston.head()
```
